# StockPhantom — Manager trust view
Top-20 go-check list + sales-vs-stock chart for any flag.

```python
import pandas as pd
alerts = pd.read_csv('../alerts.csv')
day, store = alerts['date'].max(), alerts['store_id'].iloc[0]
top = alerts[(alerts.date==day)&(alerts.store_id==store)].sort_values('rank').head(20)
top[['rank','sku_id','confidence','zero_days','expected_daily','system_on_hand','predicted_type','reason']]
```

```python
import matplotlib.pyplot as plt
fact = pd.read_parquet('../data/sales_inventory.parquet', columns=['date','store_id','sku_id','units_sold','system_on_hand'])
fact['date']=pd.to_datetime(fact['date'])
sku = top.iloc[0]['sku_id']
g = fact[(fact.store_id==store)&(fact.sku_id==sku)&(fact.date<=day)].sort_values('date').tail(60)
fig, ax1 = plt.subplots(figsize=(11,4)); ax1.bar(g['date'], g['units_sold'])
ax2=ax1.twinx(); ax2.plot(g['date'], g['system_on_hand'], color='red')
ax1.set_title(f'{store} x {sku} to {day} — blue=sales, red=system stock'); plt.xticks(rotation=30); plt.tight_layout(); plt.show()
```
